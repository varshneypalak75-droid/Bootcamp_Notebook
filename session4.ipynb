```python
# column cleanup
df.column= (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[a-z0-9]+","_",regex=True)
    .str.strip("_")
)
print(df.columns.tolist())
```

```python
df["student_id"]=df["student_id"].astype("string")
df["city"]=df["city"].astype("category")
```

```python
# missing values
missing_demo = df.copy()
missing_demo.loc[2, 'science'] = np.nan
missing_demo.loc[8, 'attendance_percent'] = np.nan
missing_demo.loc[14, 'city'] = None
print(missing_demo.isna().sum())

```

```python
science_median= missing_demo["science"].median()
attendance_median = missing_demo["attendance_percent"].median()
city_mode= missing_demo["city"].mode()[0]

missing_demo["science"]=missing_demo["science"].fillna(science_median)
missing_demo['attendance_percent']=missing_demo['attendance_percent'].fillna(attendance_median)
```

```python
practice_df=df.copy()
practice_df.loc[0,'math']=np.nan
print(practice_df["math"])
print(practice_df.isna().sum())
math_median=practice_df["math"].median()
practice_df["math"]=practice_df["math"].fillna(math_median)
print(practice_df["math"])
```

```python
# averages and some basic calculations
# derive a new column (feature creation) fro each student we will calculate their subject averages

df["average_scores"]=(
    df[['math','science','english']]
    .mean(axis=1)
    .round(2)
)
print(df[["name","math","science","english","average_scores"]].head())
```

```python
# count mean, minimun and max
top_student=df.loc[df["average_scores"].idxmax(),["name","average_scores"]]
print(top_student)
```
