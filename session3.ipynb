```python
def prepare_record(student):
    new_student=student.copy()
    new_student["name"] = new_student["name"].strip().title()
    new_student["average"] = int(sum(new_student["marks"]) / len(new_student["marks"]))
    return new_student


raw = {"name": "  riya ", "marks": [55, 63, 71]}
prepared = prepare_record(raw)

print(raw)
print(prepared)
```
