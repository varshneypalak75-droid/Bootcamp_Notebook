```python
#sota - state of the art
import pandas as pd

students = pd.DataFrame({
    "name": ["Aarav", "Riya", "Kabir", "Meera", "Dev", "Sara"],
    "study_hours": [6, 3, 8, 2, 5, 7],
    "attendance": [85, 70, 90, 60, 78, 92],
    "previous_marks": [78, 55, 88, 35, 65, 84],
    "final_marks": [82, 58, 91, 40, 70, 88]
})

students
```

```python
import numpy as np

point_a = np.array([6, 85])
point_b = np.array([3, 70])

distance = np.sqrt((point_a[0] - point_b[0])**2 + (point_a[1] - point_b[1])**2)

print("Distance =", distance)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(students["study_hours"], students["attendance"] )

for i, row in students.iterrows():
    plt.text(row["study_hours"], row["previous_marks"], row["name"])

plt.xlabel("Study Hours")
plt.ylabel("Previous Marks")
plt.title("Student Study Hours vs Previous Marks")
plt.grid(True)

plt.show()
```
